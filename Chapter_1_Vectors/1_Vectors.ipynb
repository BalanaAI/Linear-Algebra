### Plotting 2D Vector

```python
import matplotlib.pyplot as plt

v = [3, -2]

fig, axis = plt.subplots(figsize=(5, 5))      # figsize makes the figure 5 × 5 inches.

# Plot vector using quiver() function. The quiver() function is specifically designed for drawing arrows/vectors.
axis.quiver(
    0, 0, v[0], v[1],                      # starting x, starting y, vector x-component, vector y-component
    angles='xy',                           # Vector components should be interpreted using the actual x-y coordinate system
    scale_units='xy',                      # Use the coordinate system itself for determining the vector's size
    scale=1,                               # Don't shrink or enlarge the vector. Vector is drawn exactly with coordinates
    color='royalblue',                     # Vector color
    width=0.008,                           # Controls the thickness of the arrow shaft
    headwidth=5,                           # Control the size of the arrowhead
    headlength=5
)

# Make axes cross at the origin
axis.spines['left'].set_position('zero')
axis.spines['bottom'].set_position('zero')

# Hide top and right borders
axis.spines['top'].set_visible(False)
axis.spines['right'].set_visible(False)

# Set the Axis limits
axis.set_xlim(-4, 4)
axis.set_ylim(-4, 4)

# Equal scaling, that is, make one unit on the x-axis the same physical length as one unit on the y-axis
axis.set_aspect('equal')

# Add the background grid
axis.grid(True, linestyle='--', alpha=0.35)                  # alpha specify the transparency 

# Add the Labels to the coordinate axis
axis.set_xlabel('x', loc='right', fontsize=12)
axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)     # rotation=0 keeps the y horizontal instead of vertical

# Add the vector's mathematical label
axis.text(
    3.1, -2.0,                   # Specify position where the text should appear
    r'$\mathbf{v}=[3,-2]$',      # Python raw string using mathematical Latex for vector label
    fontsize=13,
    color='royalblue'
)

# Put a small black dot at the Origin
axis.scatter(0, 0,
           color='black',
           s=30,                      # s Controls the size of the dot
           zorder=5),                 # zorder makes sure the dot appears on top of other plot elements

# Display the completed figure
plt.show()
```

### Plotting a 3D Vector

```python
# Import Matplotlib Module
import matplotlib.pyplot as plt

# Define the 3D vector
v = [4, -3, 2]

# Create 3D figure
fig, axis = plt.subplots(
    figsize=(7, 7),                          # Make 7 x 7 inches Figure
    subplot_kw={'projection': '3d'}          # Make a three-dimensional coordinate system
)

# Plot the vector
axis.quiver(
    0, 0, 0,                 # Starting point x_start, y_start, and z_start
    v[0], v[1], v[2],        # Vector components dx, dy, dz
    color='royalblue',
    linewidth=3,
    arrow_length_ratio=0.12  # Controls the size of the arrowhead relative to the vector
)

# Draw 3D coordinate axes
axis_length = 5              # We are going to make our coordinate axes extend from -5 to +5 and store this in variable

# X-axis
axis.quiver(
    -axis_length, 0, 0,     # The arrow starts at -5 and moves +10. So, the axis go from (-5,0,0) to (5,0,0)
    2 * axis_length, 0, 0,
    color='black',
    linewidth=1.5,
    arrow_length_ratio=0.04
)

# Y-axis
axis.quiver(
    0, -axis_length, 0,
    0, 2 * axis_length, 0,  # The arrow starts at -5 and moves +10. So, the axis go from (0,-5,0) to (0,5,0)
    color='black',
    linewidth=1.5,
    arrow_length_ratio=0.04
)

# Z-axis
axis.quiver(
    0, 0, -axis_length,    # The arrow starts at -5 and moves upward by +10. So, the axis go from (0,0,-5) to (0,0,5)
    0, 0, 2 * axis_length,
    color='black',
    linewidth=1.5,
    arrow_length_ratio=0.04
)

# Set the x-axis, y-axis and z-axis limits
axis.set_xlim(-5, 5)         # Show x-values from −5 to +5
axis.set_ylim(-5, 5)         # Show y-values from −5 to +5
axis.set_zlim(-5, 5)         # Show z-values from −5 to +5

# Equal scaling in all three dimensions
axis.set_box_aspect([1, 1, 1])              # The x, y, and z dimensions should have equal physical proportions

# Axis labels
axis.set_xlabel('x', fontsize=14, labelpad=10)   # The labelpads adds some space between the label and the axis
axis.set_ylabel('y', fontsize=14, labelpad=10)
axis.set_zlabel('z', fontsize=14, labelpad=10)

# Mark the origin by putting a point at the origin
axis.scatter(               # scatter() creates a point
    0, 0, 0,
    color='black',
    s=45,     # s controls the size of the point
    zorder=5  # zorder controls which objects are drawn in front of others. Higher zorder means the object is placed on top
)

# Add the vector label
axis.text(
    4.5, -3.5, 2.5,                   # 3D coordinates where the text is placed
    r'$\mathbf{v} = [4, -3, 2]$',
    fontsize=14,
    color='royalblue'
)

# Turns on the 3D grid
axis.grid(True, linestyle='--', alpha=0.3)

# Set a 3D viewing angle, or Camera angle
axis.view_init(         # view_init() controls where that camera is positioned
    elev=30,            # elev controls the elevation of the camera, that is, looking at the coordinate system from above
    azim=65             # azim controls the rotation around the z-axis
)

# Set the Title of the Plot
axis.set_title(
    r'3D Vector $\mathbf{v} = [4, -3, 2]$',
    fontsize=14,
    pad=12
)

# Display the plot
plt.show()
```
