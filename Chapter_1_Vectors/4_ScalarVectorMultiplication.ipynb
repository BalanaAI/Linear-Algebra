### Scalar-Vector Multiplication
Scalar multiplication changes the length (magnitude) of a vector without changing its orientation when the scalar is positive. In other words, multiplying a vector **v** by a scalar **λ** stretches or shrinks it along its original direction. If **λ > 1**, the vector becomes longer, while if **0 < λ < 1**, it becomes shorter. The direction remains unchanged because the vector is not rotated.

When the scalar is negative **(λ < 0)**, the vector is still stretched or shrunk according to the magnitude of **λ**, but its direction is reversed. Thus, **λv** and **v** lie on the same infinitely extended straight line, but they point in opposite directions. For **λ = 0**, the vector becomes the zero vector, which has zero length and no defined direction.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define 2D vector
v = np.array([2, 3])

# Define scalars
s1 = 2
s2 = 0.3
s3 = -1.5

# Calculate the scaled vectors
v1 = s1 * v
v2 = s2 * v
v3 = s3 * v

# All vectors that will appear in the plots
vectors = [v, v1, v2, v3]

# Find the largest absolute x and y coordinates
max_x = max(abs(vector[0]) for vector in vectors)
max_y = max(abs(vector[1]) for vector in vectors)

# Add 20% margin around the largest vector
margin = 0.20

# Determine the plot limits dynamically
x_limit = max_x * (1 + margin)
y_limit = max_y * (1 + margin)

# Use the same limit for both axes. This keeps the physical scale of x and y equal
plot_limit = max(x_limit, y_limit)

# Function for plotting a vector
def plot_vector(axis, start, vector, color, label):
    axis.quiver(
        start[0],
        start[1],
        vector[0],
        vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        color=color,
        width=0.008,
        headwidth=5,
        headlength=5,
        label=label
    )

# Function for creating a complete figure
def create_vector_plot(vector, scaled_vector, title, scaled_label):

    # Create a separate figure
    fig, axis = plt.subplots(figsize=(7, 7))

    # Plot the original vector
    plot_vector(
        axis,
        [0, 0],
        vector,
        'royalblue',
        'v'
    )

    # Plot the scaled vector
    plot_vector(
        axis,
        [0, 0],
        scaled_vector,
        'gold',
        scaled_label
    )

    # Set the Title of the axis
    axis.set_title(title, pad=15, fontsize=14)

    # Legend settings
    axis.legend(loc='upper left', fontsize=11)

    # Dynamic axis limits
    axis.set_xlim(-plot_limit, plot_limit)
    axis.set_ylim(-plot_limit, plot_limit)

    # Equal scaling
    axis.set_aspect('equal')

    # Grid Setting
    axis.grid(True, linestyle='--', alpha=0.35)

    # Make x and y axes cross at the origin
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)

    # Setting Axis labels
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)

    # Mark the origin
    axis.scatter(0, 0, color='black', s=30, zorder=5)

    # Adjust layout
    plt.tight_layout()

    # Display the figure
    plt.show()
```

```python
# Case 1: λ > 1
create_vector_plot(
    v,
    v1,
    'λ > 1: Vector is Stretched',
    's₁v'
)
```

```python
# Case 2: 0 < λ < 1
create_vector_plot(
    v,
    v2,
    '0 < λ < 1: Vector is Shrunk',
    's₂v'
)
```

```python
# Case 3: λ < 0
create_vector_plot(
    v,
    v3,
    'λ < 0: Direction is Reversed',
    's₃v'
)
```
