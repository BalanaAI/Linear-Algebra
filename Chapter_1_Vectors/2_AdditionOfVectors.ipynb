### Vector Addition
To add two vectors **a** and **b**, place the tail of vector **b** at the head of vector **a**. The resulting vector, drawn from the tail of vector **a** to the head of vector **b**, represents the sum **a + b**. Let's plot the vector addition for vectors **a = [1 2]** and **b = [2 1]**. 

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2])
b = np.array([2, 1])
# Algebraic Vector Addition 
c = a + b

# Create two plots side by side for Geometric represntation of vectors and their addition
fig, (axis1, axis2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot vector a on axis1
axis1.quiver(
    0, 0, a[0], a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='royalblue',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='a'
)

# Plot vector b on axis1
axis1.quiver(
    0, 0, b[0], b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='gold',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='b'
)

axis1.set_title('Vectors a and b', pad=20)
axis1.legend(loc='upper left')

# Plot vector a on axis2
axis2.quiver(
    0, 0, a[0], a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='royalblue',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='a'
)

# plot vector b on axis2
axis2.quiver(
    a[0], a[1], b[0], b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='gold',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='b'
)

# Plot vector c = a + b on axis2
axis2.quiver(
    0, 0, c[0], c[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='lightgreen',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='c = a + b'
)

axis2.set_title('Vector addition: c = a + b', pad=20)
axis2.legend(loc='upper left')

# Common formatting for both plots
for axis in (axis1, axis2):
    # Make axes cross at the origin
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)
    
    # Set the Axis limits
    axis.set_xlim(-4, 4)
    axis.set_ylim(-4, 4)
    
    # Equal scaling, that is, make one unit on the x-axis the same physical length as one unit on the y-axis
    axis.set_aspect('equal')
    
    # Add the background grid
    axis.grid(True, linestyle='--', alpha=0.35)                  # alpha specify the transparency 
    
    # Add the Labels to the coordinate axis
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)     # rotation=0 keeps the y horizontal instead of vertical
    
    # Put a small black dot at the Origin
    axis.scatter(0, 0,
               color='black',
               s=30,                      # s Controls the size of the dot
               zorder=5),                 # zorder makes sure the dot appears on top of other plot elements

# tight_layout() automatically adjusts the spacing around and between our plots,
# so that things like titles, axis labels, tick labels, and legends don't overlap or get cut off
plt.tight_layout()
# Display the completed figure
plt.show()
```


### Vector addition is Commutative: **a** + **b** = **b** + **a**

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2])
b = np.array([2, 1])

# Create two plots side by side
fig, (axis1, axis2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot vector a on axis1
axis1.quiver(
    0, 0, a[0], a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='royalblue',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='a'
)

# Plot vector b on axis1
axis1.quiver(
    a[0], a[1], b[0], b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='gold',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='b'
)

# Plot vector a + b on axis1
axis1.quiver(
    0, 0, a[0] + b[0], a[1] + b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='lightgreen',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='a + b'
)

axis1.set_title('Vector addition: a + b', pad=20)
axis1.legend(loc='upper left')

# Plot vector b on axis2
axis2.quiver(
    0, 0, b[0], b[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='gold',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='a'
)

# plot vector a on axis2
axis2.quiver(
    b[0], b[1], a[0], a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='royalblue',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='b'
)

# Plot vector c = a + b on axis2
axis2.quiver(
    0, 0, b[0] + a[0], b[1] + a[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='lightgreen',
    width=0.008,
    headwidth=5,
    headlength=5,
    label='b + a'
)

axis2.set_title('Vector addition: b + a', pad=20)
axis2.legend(loc='upper left')

# Common formatting for both plots
for axis in (axis1, axis2):
    # Make axes cross at the origin
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)
    
    # Set the Axis limits
    axis.set_xlim(-4, 4)
    axis.set_ylim(-4, 4)
    
    # Equal scaling, that is, make one unit on the x-axis the same physical length as one unit on the y-axis
    axis.set_aspect('equal')
    
    # Add the background grid
    axis.grid(True, linestyle='--', alpha=0.35)                  # alpha specify the transparency 
    
    # Add the Labels to the coordinate axis
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)     # rotation=0 keeps the y horizontal instead of vertical
    
    # Put a small black dot at the Origin
    axis.scatter(0, 0,
               color='black',
               s=30,                      # s Controls the size of the dot
               zorder=5),                 # zorder makes sure the dot appears on top of other plot elements

# tight_layout() automatically adjusts the spacing around and between our plots,
# so that things like titles, axis labels, tick labels, and legends don't overlap or get cut off
plt.tight_layout()
# Display the completed figure
plt.show()
```
