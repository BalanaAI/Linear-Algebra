### Vector Subtraction
Vector subtraction can be understood in two equivalent ways. The first is to turn subtraction into addition by multiplying the vector being subtracted by −1. This changes the sign of every component and flips the vector by 180°. For example, the vector **[2 1]** becomes **[−2 −1]**. We can then add this reversed vector to the other vector, just as we do in vector addition.

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2])
b = np.array([2, 1])
# Algebraic Vector Subtraction
b = -1 * b
c = a + b

# Function for plotting a vector
def plot_vector(axis, start, vector, color, label):
    axis.quiver(
        start[0], start[1],
        vector[0], vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        color=color,
        width=0.008,
        headwidth=5,
        headlength=5,
        label=label
    )

# Create two plots side by side for Geometric representation of vectors and their subtraction
fig, (axis1, axis2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot vectors a and b on axis1
plot_vector(axis1, [0, 0], a, 'royalblue', 'a')
plot_vector(axis1, [0, 0], b, 'gold', '-b')

axis1.set_title('Vectors a and -b', pad=20)
axis1.legend(loc='upper left')

# Plot vector a, b and c on axis2
plot_vector(axis2, [0, 0], a, 'royalblue', 'a')
plot_vector(axis2, a, b, 'gold', 'b')
plot_vector(axis2, [0, 0], c, 'lightgreen', 'c = a - b')

axis2.set_title('Vector Subtraction: c = a - b', pad=20)
axis2.legend(loc='upper left')

# Common formatting for both plots
for axis in (axis1, axis2):
    # Make axes cross at the origin
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)
    
    # Set the Axis limits
    axis.set_xlim(-4, 4)
    axis.set_ylim(-4, 4)
    
    # Equal scaling, that is, make one unit on the x-axis the same physical length as one unit on the y-axis
    axis.set_aspect('equal')
    
    # Add the background grid
    axis.grid(True, linestyle='--', alpha=0.35)                  # alpha specify the transparency 
    
    # Add the Labels to the coordinate axis
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)     # rotation=0 keeps the y horizontal instead of vertical
    
    # Put a small black dot at the Origin
    axis.scatter(0, 0,
               color='black',
               s=30,                      # s Controls the size of the dot
               zorder=5),                 # zorder makes sure the dot appears on top of other plot elements

# tight_layout() automatically adjusts the spacing around and between our plots,
# so that things like titles, axis labels, tick labels, and legends don't overlap or get cut off
plt.tight_layout()
# Display the completed figure
plt.show()
```


The second approach is to keep both vectors in their standard positions, with their tails at the origin. Then draw a vector from the **head of the vector being subtracted** to the **head of the other vector**. This new vector represents the difference between the two vectors. Although it is not in standard position, its position does not affect the vector itself; what matters is its magnitude and direction.

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2])
b = np.array([2, 1])

# Function for plotting a vector
def plot_vector(axis, start, vector, color, label):
    axis.quiver(
        start[0], start[1],
        vector[0], vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        color=color,
        width=0.008,
        headwidth=5,
        headlength=5,
        label=label
    )

# Create two plots side by side
fig, (axis1, axis2) = plt.subplots(1, 2, figsize=(10, 5))

# Plot vector a and b on axis1
plot_vector(axis1, [0, 0], a, 'royalblue', 'a')
plot_vector(axis1, [0, 0], b, 'gold', 'b')

axis1.set_title('Vectors a and b', pad=20)
axis1.legend(loc='upper left')

# Plot vector a, b and a-b on axis2
plot_vector(axis2, [0, 0], a, 'royalblue', 'a')
plot_vector(axis2, [0, 0], b, 'gold', 'b')
plot_vector(axis2, b, a-b, 'lightgreen', 'a - b')

axis2.set_title('Vector Subtraction: a - b', pad=20)
axis2.legend(loc='upper left')

# Common formatting for both plots
for axis in (axis1, axis2):
    # Make axes cross at the origin
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)
    
    # Set the Axis limits
    axis.set_xlim(-4, 4)
    axis.set_ylim(-4, 4)
    
    # Equal scaling, that is, make one unit on the x-axis the same physical length as one unit on the y-axis
    axis.set_aspect('equal')
    
    # Add the background grid
    axis.grid(True, linestyle='--', alpha=0.35)                  # alpha specify the transparency 
    
    # Add the Labels to the coordinate axis
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)     # rotation=0 keeps the y horizontal instead of vertical
    
    # Put a small black dot at the Origin
    axis.scatter(0, 0,
               color='black',
               s=30,                      # s Controls the size of the dot
               zorder=5),                 # zorder makes sure the dot appears on top of other plot elements

# tight_layout() automatically adjusts the spacing around and between our plots,
# so that things like titles, axis labels, tick labels, and legends don't overlap or get cut off
plt.tight_layout()
# Display the completed figure
plt.show()
```

Note that **vector subtraction** is valid only when the vectors have the same dimensionality.
