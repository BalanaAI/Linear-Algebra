### Code Challenge
Create a 2D vector v, 10 scalars that are drawn at random from a normal (Gaussian) distribution, and plot all 10 scalar vector multiplications on top of each other. What do you notice?

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the vector
v = np.array([2, 1])

# Generate 10 random scalars from a standard normal distribution
scalars = np.random.normal(0, 1, 10)

# Starting point
start = np.array([0, 0])

# Calculate all scalar-vector multiples
vectors = [scalar * v for scalar in scalars]

# Find the largest coordinate among all vector endpoints
max_coordinate = max(np.max(np.abs(vector)) for vector in vectors)

# Add 20% margin
plot_limit = max_coordinate * 1.20

# Prevent a zero-sized plot
if plot_limit == 0:
    plot_limit = 1

# Create the plot
fig, axis = plt.subplots(figsize=(6, 6))

# Generate 10 different colors for 10 different scalar-vector multiples
colors = plt.cm.tab10(np.arange(10))

# Plot all 10 scalar multiples
for scalar, vector, color in zip(scalars, vectors, colors):
    # Plot the vector
    axis.quiver(
        start[0],
        start[1],
        vector[0],
        vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        color=color,
        width=0.008,
        label=f'{scalar:.2f}v'
    )

# Set dynamic axis limits
axis.set_xlim(-plot_limit, plot_limit)
axis.set_ylim(-plot_limit, plot_limit)

# Set equal scaling
axis.set_aspect('equal')

# Add grid
axis.grid(True, linestyle='--', alpha=0.35)

# Make axes cross at the origin
axis.spines['left'].set_position('zero')
axis.spines['bottom'].set_position('zero')
# Hide top and right borders
axis.spines['top'].set_visible(False)
axis.spines['right'].set_visible(False)

# Add axis labels
axis.set_xlabel('x', loc='right')
axis.set_ylabel('y', loc='top', rotation=0)

# Add legend
axis.legend(loc='upper left', fontsize=9)

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```

The main thing we noticed is that **scalar multiplication changes the magnitude of a vector but does not change its line of direction**.

In geometric terms, **scalar multiplication stretches, shrinks, or reverses a vector, but it does not change the line on which the vector lies**.
