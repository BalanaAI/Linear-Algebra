First let's write the functions to plot the vectors

```python
import numpy as np
import matplotlib.pyplot as plt

# Function for plotting a vector
def plot_vector(axis, start, vector, color='royalblue', label='v'):
    # Plot the vector
    axis.quiver(
        start[0],
        start[1],
        vector[0],
        vector[1],
        angles='xy',           # Interpret the vector components, (x,y), using the actual x-y coordinate system
        scale_units='xy',
        scale=1,
        color=color,
        # width, headwidth, and headlength control the visual appearance of the arrow
        width=0.008,
        headwidth=5,
        headlength=5,
        label=label
    )

    # Calculate the vector endpoint
    end = start + vector

    # Add starting point coordinates
    axis.annotate(                       # annotate() is used to place text associated with a particular point on the plot
        f'({start[0]:g}, {start[1]:g})', # g specifies general number formatting like 2 instead of 2.000000
        xy=(start[0], start[1]),         # Specifies the location where the annotation belongs
        xytext=(8, -12),                 # These offset values controls where the text appears relative to the point
        textcoords='offset points',      # Treat xytext as an offset measured in points
        fontsize=11,
        color='black'
    )

    # Add ending point coordinates
    axis.annotate(
        f'({end[0]:g}, {end[1]:g})',
        xy=(end[0], end[1]),
        xytext=(8, 8),
        textcoords='offset points',
        fontsize=11,
        color='black'
    )


# Function for creating a complete figure
def create_vector_plot(vector, starting_point, title, color='royalblue'):
    # Calculate the endpoint of the vector
    endpoint = starting_point + vector
    
    # Determine how large the plotting area needs to be. That is, the largest coordinate that must be visible
    max_coordinate = max(
        abs(starting_point[0]),
        abs(starting_point[1]),
        abs(endpoint[0]),
        abs(endpoint[1])
    )
    
    # Add 20% margin
    margin = 0.20
    # Determine the plot limits dynamically
    plot_limit = max_coordinate * (1 + margin)

    # Prevent a zero-sized plot
    if plot_limit == 0:
        # Set minimum plot size
        plot_limit = 1

    # Create figure
    fig, axis = plt.subplots(figsize=(5, 5))   # Create a Square figure of 5 x 5 inches

    # Plot the vector and coordinates
    plot_vector(
        axis,
        starting_point,
        vector,
        color=color,
        label='v'
    )

    # Set title of the plot
    axis.set_title(title, pad=15, fontsize=14)

    # Set symmetric axis limits. This keeps (0, 0) exactly at the center
    axis.set_xlim(-plot_limit, plot_limit)
    axis.set_ylim(-plot_limit, plot_limit)
    
    # Equal physical scaling for x and y. That is, make one unit in x have the same physical length as one unit in y
    axis.set_aspect('equal')

    # Grid Settings
    axis.grid(True, linestyle='--', alpha=0.35)
    
    # Make x and y axes cross at the origin
    # That is, move the vertical axis to x = 0 and move the horizontal axis to y = 0
    axis.spines['left'].set_position('zero')
    axis.spines['bottom'].set_position('zero')

    # Hide top and right borders
    axis.spines['top'].set_visible(False)
    axis.spines['right'].set_visible(False)

    # Axis labels
    axis.set_xlabel('x', loc='right', fontsize=12)
    axis.set_ylabel('y', loc='top', fontsize=12, rotation=0)

    # Mark the origin
    axis.scatter(0, 0, color='black', s=30, zorder=5)
    
    # Legend Settings
    axis.legend(loc='upper left', fontsize=11)
    
    # Adjust layout
    plt.tight_layout()
    
    # Display
    plt.show()
```

```python
# Plot the vector [2 2] (0, 0)
v = np.array([2, 2])
s_point = np.array([0, 0])

create_vector_plot(
    v,
    s_point,
    'v = [2, 2]',
    'green'
)
```
