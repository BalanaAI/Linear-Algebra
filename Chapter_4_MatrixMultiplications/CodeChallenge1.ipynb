### Code Challenge - Matrix Multiplication by Layering
In this code challenge we have to implement matrix multiplication by using the **Layering** approach with the help of **for** loops and **outer product**.<br>
**Note** that the layer-wise approach look at the matrix muliplication like **"Each matching column of A and row of B produces one entire matrix layer. Add all those layers together."**

```python
# Import Numpy Module
import numpy as np
```

```python
# Create sizes of matrices
m = 4
n = 6
```

```python
# Generate two matrices (A, B)
A = np.random.randint(-9, 9, (m, n))           # A = 4 x 6
B = np.random.randint(-9, 9, (n, m))           # B = 6 x 4

print("A =\n", A)
print(" ")
print("B =\n", B)
```

```python
# Build the Product-matrix Layer-wise (Using for loop)
C1 = np.zeros((m, m))       # Remember that zeros() function take the size in the form of tuple
# Create a Layer matrix, using outer product, at each iteration and add to the resultant matrix
for i in range(n):
    C1 = C1 + np.outer(A[:, i], B[i, :])
print("C1 =\n", C1)
```

```python
# Implement the matrix multiplication directly
C2 = np.matmul(A, B)
print("C2 =\n", C2)
```

```python
# Compare the results
print(C1 - C2)
```
