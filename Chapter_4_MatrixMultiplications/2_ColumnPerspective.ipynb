### **Matrix Multiplication - Column Perspective**
In the **column perspective of matrix multiplication**, each column of the product matrix is formed as a linear combination of the columns of the left matrix. The coefficients, or weights, used in this combination are taken from the corresponding column of the right matrix. Specifically, the first column of the product is obtained by weighting the columns of the left matrix using the entries of the first column of the right matrix. Similarly, the second column of the product is obtained using the entries of the second column of the right matrix as weights. This process continues for every column of the right matrix, with each column determining the weights used to construct the corresponding column of the product matrix.

```python
# Import Numpy Module
import numpy as np
```

```python
# Create sizes of matrices
m = 3
n = 4
```

```python
# Generate two matrices (A, B)
A = np.random.randint(-9, 9, (m, n))           # A = 3 x 4
B = np.random.randint(-9, 9, (n, m))           # B = 4 x 3

print("A =\n", A)
print(" ")
print("B =\n", B)
```

<br>

Product-matrix **C** as Linear Combination of Columns (using for loop)

```python
# Create an empty product matrix
C = np.zeros((A.shape[0], B.shape[1]))
# Column perspective of matrix multiplication
for j in range(C.shape[1]):            # Iterate over the columns of the product matrix C
    for k in range(A.shape[1]):        # Iiterate over the columns of A, while also selecting the corresponding row of B
        C[:, j] += A[:, k] * B[k,j]    # A[:, k] selects the k-th column of A, and B[k, j] provides the weight for that column
print("C =\n", C)
```

```python
# Implement the matrix multiplication directly
D = A@B
print("D =\n", D)
```

```python
# Compare the results
print(C - D)
```
