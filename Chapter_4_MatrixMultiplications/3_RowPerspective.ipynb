### **Matrix Multiplication - Row Perspective**
In the **row perspective of matrix multiplication**, each row of the product matrix is formed as a linear combination of the rows of the right matrix. The coefficients, or weights, used in this combination are taken from the corresponding row of the left matrix. Specifically, the first row of the product is obtained by weighting the rows of the right matrix using the entries of the first row of the left matrix. Similarly, the second row of the product is obtained using the entries of the second row of the left matrix as weights. This process continues for every row of the left matrix, with each row determining the weights used to construct the corresponding row of the product matrix.

```python
# Import Numpy Module
import numpy as np
```

```python
# Create sizes of matrices
m = 3
n = 4
```

```python
# Generate two matrices (A, B)
A = np.random.randint(-9, 9, (m, n))           # A = 3 x 4
B = np.random.randint(-9, 9, (n, m))           # B = 4 x 3

print("A =\n", A)
print(" ")
print("B =\n", B)
```

<br>

Product-matrix **C** as Linear Combination of Rows (using for loop)

```python
# Create an empty product matrix C = AB
C = np.zeros((A.shape[0], B.shape[1]))
# Row perspective of matrix multiplication
for j in range(C.shape[0]):            # Iterate over each row of the product matrix C
    for k in range(B.shape[0]):        # Iterate over each row of B
        C[j, :] += B[k, :] * A[j,k]    # Add the k-th row of B, weighted by A[j, k], to row j of C
print("C =\n", C)
```

```python
# Implement the matrix multiplication directly
D = A@B
print("D =\n", D)
```

```python
# Compare the results
print(C - D)
```
