### **Standard Matrix Multiplication**
If **A** is an ***m×n*** matrix and **B** is an ***n×p*** matrix, their product **C = AB** is an ***m×p*** matrix. For matrix multiplication to be valid, the inner dimensions (number of columns of **A** and number of rows of **B**) must match, while the outer dimensions (number of rows of **A** and number of columns of **B**) determine the dimensions of the resulting matrix.

$$
A_{m \times n} B_{n \times p} = C_{m \times p}
$$

$$
\underbrace{m}_{\text{Outer}}
\times
\underbrace{n}_{\text{Inner}}
\quad
\underbrace{n}_{\text{Inner}}
\times
\underbrace{p}_{\text{Outer}}
$$

Thus, the key condition for matrix multiplication is that **the number of columns in the first matrix must equal the number of rows in the second matrix**. In general, matrix multiplication is **not commutative**; that is, **AB ≠ BA**.

```python
# Import Numpy Module
import numpy as np
```

```python
# Variables for matrix dimensions
m = 4
n = 3
p = 6
```

```python
# Create three matrices of random integers
A = np.random.randint(-9, 9, (m,n))    # 4 x 3 matrix
B = np.random.randint(-9, 9, (n,p))    # 3 x 6 matrix
C = np.random.randint(-9, 9, (m,p))    # 4 x 6 matrix

print("A =\n", A)
print(" ")
print("B =\n", B)
print(" ")
print("C =\n", C)
```
