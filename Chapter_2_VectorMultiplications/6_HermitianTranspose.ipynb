### Hermitian Transpose
A **complex conjugate** changes the sign of the imaginary part of a complex number: $a + bi \;\longrightarrow\; a - bi$; geometrically, it reflects the number across the real axis.

The **Hermitian transpose** applies this conjugation to every element of a vector or matrix **and then transposes it** (swaps rows and columns). For example, if 

$$
\mathbf{a} =
\begin{bmatrix}
1+i\\
2-i
\end{bmatrix}
\quad\Longrightarrow\quad
\mathbf{a}^H =
\begin{bmatrix}
1-i & 2+i
\end{bmatrix}
$$

We use the Hermitian transpose mainly when working with complex vectors and matrices, especially to calculate inner products such as $\mathbf{a^{H}b}$.


```python
# Import Numpy Module
import numpy as np
```

The **norm** or **magnitude** of complex number is its distance from the origin. Let's see how to calculate the norm or magnitude of complex number in Python.

```python
# Create a Complex Number
z = complex(3,4)
print(f"Complex number z = {z}")

# Magnitude of the Complex Number z using Numpy norm() Function
# The norm() function takes the complex conjugate
z_norm1 = np.linalg.norm(z)
print(f"Magnitude of z = {z_norm1}")  # 5

# Magnitude of Complex Number z using transpose() function
# The transpose() function does not take the complex conjugate
z_norm2 = np.sqrt(np.transpose(z) * z)
print(f"Magnitude of z = {z_norm2.real}")  # Incorrect result will be produced

# Magnitude of Complex Number z using Hermitian transpose procedure
# The Hermitian transpose is to first take the conjugate and then transpose
z_norm3 = np.sqrt(np.transpose(z.conjugate()) * z)
print(f"Magnitude of z = {z_norm3.real}")
```

Now let's see how to take transpose of complex number vectors

```python
# Create a Complex Vector
v = np.array( [ 3, 4j, 5+2j, complex(2,-5) ] )
print( v.T )                          # This is the simple transpose; it does not take the complex conjugate
print( np.transpose(v) )              # This is the simple transpose; it does not take the complex conjugate
print( np.transpose(v.conjugate()) )  # This is the Hermitian transpose, and it takes the complex conjugate.
```

So, when working with **complex vectors**, we must use the **Hermitian transpose** rather than the simple transpose.
