### Code Challenge - Is the Dot Product Commutative?
### a.b == b.a

- Generate two 100-elements random row vectors, and compute dot product a with b & b with a
- Generate two 2-elements integer row vectors, and compute dot product a with b & b with a

```python
import numpy as np
```

```python
# Dot Products of Random Number Vectors
a = np.random.rand(100)
b = np.random.randn(100)

a_dot_b = np.dot(a, b)
b_dot_a = np.dot(b, a)

print("a_dot_b = ", a_dot_b)
print("b_dot_a = ", b_dot_a)
```

```python
# Dot Products of Random Integer Vectors
a = np.random.randint(1, 10, 2)
b = np.random.randint(1, 10, 2)

a_dot_b = np.dot(a, b)
b_dot_a = np.dot(b, a)

print("a = ")
print(a)
print(" ")
print("b = ")
print(b)
print(" ")
print("a_dot_b = ", a_dot_b)
print("b_dot_a = ", b_dot_a)
```
