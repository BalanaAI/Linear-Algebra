### Code Challenge - Cauchy-Schwarz inequality
The Cauchy-Schwarz inequality is that the magnitude of the dot product between two vectors is less than or equal to the product of the norms of the two vectors.

$$
\boldsymbol{|a^T b| \leq \|a\| \|b\|}
$$

In this coding challenge, create three vectors, with one vector being dependent on the other two, and use them to illustrate the Cauchy–Schwarz inequality.

```python
# Import Numpy Module
import numpy as np
```

```python
# Create 3 vectors
a = np.random.randn(5)        # Independent vector
b = np.random.randn(5)        # Independent vector
c = np.random.randn() * a     # Vector c is linearlly dependent on vector a
```

```python
# Case-1: Cauchy–Schwarz Inequality for Independent Vectors
#
# LHS of the Cauchy–Schwarz Inequality
lhs_aTb = np.dot(a, b)

# RHS of the Cauchy–Schwarz Inequality
rhs_ma_into_mb = np.linalg.norm(a) * np.linalg.norm(b)

print(f"[{abs(lhs_aTb):.4f}, {rhs_ma_into_mb:.4f}]")
```

```python
# Case-2: Cauchy–Schwarz Inequality for Dependent Vectors
#
# LHS of the Cauchy–Schwarz Inequality
lhs_aTc = np.dot(a, c)

# RHS of the Cauchy–Schwarz Inequality
rhs_ma_into_mc = np.linalg.norm(a) * np.linalg.norm(c)

print(f"[{abs(lhs_aTc):.4f}, {rhs_ma_into_mc:.4f}]")
```

Thus, we have illustrated through the code that the Cauchy–Schwarz inequality becomes an **equality** when the two vectors point in the same or exactly opposite direction (i.e., when they are linearly dependent).

The inequality is **strict (i.e., the left-hand side is strictly less than the right-hand side)** when the vectors are not parallel, meaning that they point in different directions.
