### Code Challenge - Dot Product Sign and Scalar Multiplication
In this code challenge we have to test whether the **Dot Product Sign** is invariant to scalar multiplication or not. So,
- Generate two vectors (R3)
- Generate two scalars
- Compute the Dot Product between vectors
- Compute the Dot Product between the scaled vectors

```python
import numpy as np
```

```python
# Generate two vectors (R3)
v1 = np.array([-3, 4, 5])
v2 = np.array([3, 6, -3])

# Generate two scalar
s1 = 2
s2 = 3
```

```python
# Compute Dot Product between the Original Vectors
print('Original Vectors Dot Product: ', np.dot(v1, v2))
```

Here the dot product between original vectors is zero. It means that they are orthogonal to each other. In other words, these vector v1 and v2 meet at right angle.

```python
# Compute Dot Product between the Scaled Vectors
print('Scaled Vectors Dot Product: ', np.dot(s1*v1, s2*v2))
```

The dot product between scaled vectors is also zero. It means when the dot product between two vectors is zero, then the dot product between scaled versions of those two vectors is also zero.<br>
Geometrically it means when the original vectors are orthogonal to each other then their scaled versions are also orthogonal to each other.

```python
# Let's make the two vectors non-orthogonal to each other by changing vector v1
v1 = np.array([-3, 4, 6])
```

```python
# Compute Dot Product between the Original Non-orthogonal Vectors
print('Original Vectors Dot Product: ', np.dot(v1, v2))
```

```python
# Compute Dot Product between the Scaled Non-orthogonal Vectors
print('Scaled Vectors Dot Product: ', np.dot(s1*v1, s2*v2))
```

Here in case of Non-orthogonal vectors the dot product of original vectors is -3 while the dot product of scaled vectors is -18. This also make sense both algebraically and geometrically.<br>
Let's now change the sign of one of the scalar from positive to negative.

```python
s1 = -2
```

```python
# Compute Dot Product between the Scaled Non-orthogonal Vectors
print('Scaled Vectors Dot Product: ', np.dot(s1*v1, s2*v2))
```

Now the dot product of the scaled vectors is positive, 18. So, the original dot product was negative. It means that vectors v1 and v2 are meeting at an **obtuse angle**. Then when we scaled vector v1 by -2 it changed the direction of v1 by 180 degree and now it is meeting vector v2 at an **acute angle** and their dot product becomes positive.<br>
So, the answer to the main question is that **the Dot Product Sign is not invariant to scalar multiplication**. That means that the scalar multiplication has effect on the sign of the dot product.
