## Vector Multiplications
There are four common ways to multiply a pair of vectors:
- **Dot Product**
- **Outer Product**
- **Element-wise Multiplication (Hadamard Multiplication)** and
- **Cross Product**

### **Dot Product**
The dot product produces a single numerical value that describes the relationship between two vectors. The dot product is also called **inner product** or **scalar product**. To calculate the dot product, multiply the corresponding elements of the two vectors and then add all the resulting products together. Following are the ways to compute the dot product in Python.

```python
# Import the Numpy and Matplotlib.pyplot Modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Method-1
v1 = np.array([1, 2, 3, 4, 5, 6])
v2 = np.array([0, -4, -3, 6, 5, -2])

dp1 = sum(np.multiply(v1, v2))
```

```python
# Method-2
dp2 = np.dot(v1, v2)
```

```python
# Method-3
dp3 = np.matmul(v1, v2)
```

```python
# Method-4
dp4 = 0
# Loop over the vectors
for i in range(v1.size):
    # Multiply the corresponding elements of v1 & v2 and then sum
    dp4 += v1[i] * v2[i]
```

```python
print(dp1, dp2, dp3, dp4)
```


### **Properties of Dot Product**
#### **Associative Property**
In general, the associative property does not apply to the dot product of vectors. In other words,

$$
\mathbf{u}^T(\mathbf{v}^T\mathbf{w}) \neq (\mathbf{u}^T\mathbf{v})^T\mathbf{w}
$$


However, there are two special cases in which the associative property can hold:
- Any one vector is the zero vector (subject to the dimensions being compatible).
- All vectors are identical, such that **u==v==w**.

```python
# Create random vectors of integers
n = 5
u = np.random.randint(0, 9, n)
v = np.random.randint(0, 9, n)
w = np.random.randint(0, 9, n)

# The LHS and RHS of the above Equation
LHS = np.dot(u, np.dot(v, w))
RHS = np.dot(np.dot(u, v), w)

print("LHS =", LHS)
print("RHS =", RHS)
```

Here we can see that the LHS is not equal to RHS, and thus the Associative Property for vectors dot product does not hold.

```python
# Special Case-1: Any one vector is the zero vector (subject to the dimensions being compatible)
u = np.zeros(5, dtype=int)

# The LHS and RHS of the above Equation
LHS = np.dot(u, np.dot(v, w))
RHS = np.dot(np.dot(u, v), w)

print("LHS =", LHS)
print("RHS =", RHS)
```

```python
# Special Case-2: All vectors are identical, such that u==v==w.
u = np.random.randint(0, 9, n)
v = u
w = u
print("u =", u, "\nv =", v, "\nw =", w)
print(" ")

# The LHS and RHS of the Equation
LHS = np.dot(u, np.dot(v, w))
RHS = np.dot(np.dot(u, v), w)

print("LHS =", LHS)
print("RHS =", RHS)
```

#### **Distributive Property**
The dot product is distributive over vector addition. This means that the dot product of a vector with the sum of two vectors is equal to the sum of the individual dot products.

Mathematically:

$$
\mathbf{u}^T(\mathbf{v}+\mathbf{w}) = \mathbf{u}^T\mathbf{v} + \mathbf{u}^T\mathbf{w}
$$

The distributive property always holds for **compatible vectors**.

```python
# Create 3 random vectors of integers
n = 5     # Number of elements in vector
u = np.random.randint(0, 9, n)
v = np.random.randint(0, 9, n)
w = np.random.randint(0, 9, n)

# The LHS and RHS of the Equation
LHS = np.dot(u, (v + w))
RHS = np.dot(u, v) + np.dot(u, w)

print("u =", u, "\nv =", v, "\nw =", w)
print(" ")

print("LHS =", LHS)
print("RHS =", RHS)
```

Here, we can see that **LHS** is equal to **RHS**. Thus **distributive property** holds for vectors dot product. 

#### **Commutative Property**
The dot product is commutative, meaning that switching the order of the two vectors does not change the result:

$$
a^{T}b=b^{T}a
$$


This is because the dot product is calculated by multiplying corresponding components of the two vectors and then adding those products together. Since ordinary scalar multiplication is commutative—$a_i b_i = b_i a_i$—each component-wise multiplication gives the same result regardless of which vector comes first. Therefore, swapping the vectors does not affect the final dot product.

```python
# Create 2 random vectors of integers
n = 5     # Number of elements in vector
a = np.random.randint(0, 9, n)
b = np.random.randint(0, 9, n)

# The LHS and RHS of the Equation
LHS = np.dot(a, b)
RHS = np.dot(b, a)

print("a =", a, "\nb =", b)
print(" ")

print("LHS =", LHS)
print("RHS =", RHS)
```

This illustrates that the vector dot product is **commutative**.


### **Geometric Perspective of Vector Dot Product**
Geometrically, the dot product measures the **alignment between two vectors**. It is determined by the lengths of the two vectors and by the angle between them:

$$
a^T b = \|\mathbf{a}\|\,\|\mathbf{b}\|\cos(\theta_{ab})
$$


In intuitive terms, **the more closely two vectors point in the same direction, the larger their dot product**. The cosine of the angle captures this alignment, while the vector lengths determine the scale of the result.

**Note** that the geometric and algebraic formulae are simply different ways of expressing the same concept.

```python
# Define two 3D vectors
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])

# Compute the angle (radians) between vectors v1 & v2
angle = np.arccos( np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2)) )
print("Angle between v1 & v2 =", angle, "radians")
```

```python
# Plot the Vectors v1 & v2
# 
# Create a 3D figure
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

# Draw the vectors from the origin
ax.quiver(0, 0, 0, *v1, color='royalblue', linewidth=2.5, arrow_length_ratio=0.12, label='v1')
ax.quiver(0, 0, 0, *v2, color='green', linewidth=2.5, arrow_length_ratio=0.12, label='v2')

# Label the vectors
ax.text(*v1, '  v1', color='royalblue', fontsize=12, fontweight='bold')
ax.text(*v2, '  v2', color='green', fontsize=12, fontweight='bold')

# Mark the origin
ax.scatter(0, 0, 0, color='black', s=40)

# Axis labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Set limits according to vector lengths
vectors = [v1, v2]
max_range = np.max(np.abs(vectors))
margin = 0.25 * max_range
limit = max_range + margin

ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_zlim(-limit, limit)

# Equal aspect ratio
ax.set_box_aspect([1, 1, 1])

# Set the 3D viewing angle
ax.view_init(elev=15.1, azim=15.9)

# Grid and legend
ax.grid(True)
ax.legend()

plt.show()
```

```python
# Equivalence of Algebraic and Geometric Dot Product formulae

# Define two vectors
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])

# Algebraic Dot Product of v1 & v2
dp_a = np.dot(v1, v2)

# Geometric Dot Product of v1 & v2
dp_g = np.linalg.norm(v1) * np.linalg.norm(v2) * np.cos(angle)      # 'angle' was calculated previously

# Print bot Dot Product
print(f"Algebraic Dot Product = {dp_a:.2f}")
print(f"Geometric Dot Product = {dp_g:.2f}")
```
