### Vector Length (Norm)
The length or norm of a vector tells us how far the vector extends from its starting point to its endpoint. In other words, it measures the vector’s **magnitude or size**, regardless of its direction. For a vector a=($a_1,a_2,…,a_n$), its Euclidean norm is:


$$
\|\mathbf{a}\| = \sqrt{a_1^2 + a_2^2 + \cdots + a_n^2}
$$

Intuitively, it is the **straight-line distance represented by the vector**.

```python
# Import the Numpy Module
import numpy as np
```

```python
# Methods to calculate Vector Length (Norm)
# Define a vector v
v = np.array([ 1, 2, 3, 4, 5, 6 ])

# Method-1
vl1 = np.sqrt( sum(np.multiply(v, v)) )

# Method-2
vl2 = np.sqrt( np.matmul(v, v) )

# Method-3
dp = 0
# Loop over the vector
for i in range(v.size):
    # Multiply each element of vector v with itself
    dp += v[i] * v[i]
vl3 = np.sqrt(dp)

# Method-4
vl4 = np.sqrt( np.dot(v, v) )

# method-5: Take the norm
vl5 = np.linalg.norm(v)

print(vl1,vl2, vl3, vl4, vl5)
```
