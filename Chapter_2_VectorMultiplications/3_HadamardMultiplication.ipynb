### Element-wise (Hadamard) Muliplication
The Hadamard (element-wise) product of two vectors is obtained by multiplying their corresponding components. Unlike the dot product, it produces a vector of the same dimension:

$$
\mathbf{a} \odot \mathbf{b}
=
[a_1b_1,\; a_2b_2,\; \ldots,\; a_nb_n]
$$


```python
# Import Numpy Module
import numpy as np
```

```python
# Create 2 vectors
a = [ 1, 3, 5 ]
b = [ 3, 4, 2 ]
```

```python
# Method-1: Hadmard Vector Mulitplication
resultant_vector = np.multiply(a, b)
print("Hadamard Vector Product =", resultant_vector)
```

```python
# Method-2: Hadmard Vector Mulitplication
resultant_vector = np.array(a) * np.array(b)
print("Hadamard Vector Product =", resultant_vector)
```
