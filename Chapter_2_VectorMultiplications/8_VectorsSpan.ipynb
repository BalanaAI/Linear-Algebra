### Vectors Span

```python
import numpy as np
import matplotlib.pyplot as plt

# Vectors Set for Span
S1 = np.array([1, 1, 0])
S2 = np.array([1, 7, 0])

# Vectors v and w
v = np.array([1, 2, 0])    # Vector v is in the span
w = np.array([3, 2, 1])    # Vector w is not

# Plot the Span and Vectors
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection="3d")

# Plane
x = np.linspace(-1, 4, 2)
y = np.linspace(-1, 8, 2)

X, Y = np.meshgrid(x, y)
Z = np.zeros_like(X)

ax.plot_surface(
    X, Y, Z,
    color="lightblue",
    alpha=0.8,
    edgecolor="none",
    shade=False
)

# Utility Function to Draw Vectors
def draw_vector(vec, color, label):
    # Black outline
    ax.quiver(
        0, 0, 0,
        vec[0], vec[1], vec[2],
        color="black",
        linewidth=6,
        arrow_length_ratio=0.10
    )

    # Colored vector
    ax.quiver(
        0, 0, 0,
        vec[0], vec[1], vec[2],
        color=color,
        linewidth=3.5,
        arrow_length_ratio=0.10
    )

    # Endpoint
    ax.scatter(
        vec[0], vec[1], vec[2],
        color=color,
        edgecolor="black",
        linewidth=1,
        s=70
    )

    # Vector label
    ax.text(
        vec[0],
        vec[1],
        vec[2],
        f"{label}",
        color=color,
        fontsize=13,
        weight="bold"
    )


# Draw vectors
draw_vector(S1, "red",    r"$S_1$")
draw_vector(S2, "blue",   r"$S_2$")
draw_vector(v,  "green",  r"$v$")
draw_vector(w,  "purple", r"$w$")

# Origin
ax.scatter(0, 0, 0, color="black", s=80)

# Axes Settings
ax.set_xlabel("x", fontsize=13, labelpad=5)
ax.set_ylabel("y", fontsize=13, labelpad=5)
ax.set_zlabel("z", fontsize=13, labelpad=5)

ax.set_xlim(-2, 5)
ax.set_ylim(-2, 9)
ax.set_zlim(-2, 4)

# Equal Aspect Ratio
ax.set_box_aspect((
    5 - (-2),   # x range
    9 - (-2),   # y range
    4 - (-2)    # z range
))

# Camera Settings
ax.view_init(elev=3, azim=-145)

# Plot Title
# text2D uses coordinates relative to the actual axes:
# x = 0.50  -> center
# y = 0.82  -> close to the plot
ax.text2D(
    0.50,
    0.82,
   r"$\boldsymbol{\mathrm{Vectors\ and\ span}(S_1,S_2)}$",
    transform=ax.transAxes,
    ha="center",
    va="center",
    fontsize=15,
    weight="bold"
)

# LAYOUT
# Keep the axes large and prevent Matplotlib from creating
# a large empty region around the 3D plot.
plt.subplots_adjust(left=0.03, right=0.97, bottom=0.03, top=0.95)

# Display the plot
plt.show()
```
