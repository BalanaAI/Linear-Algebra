### Code Challenge - Dot Products with Unit Vectors
The goal of this challenge is to gain experience in coding using the Dot Product and Normalizing Vectors (creating Unit Vectors along some direction). So, in this challenge you have to do:
- Create two random-integer vectors (R4)
- Compute the Lengths of the individual vectors, and magnitude of their dot product
- Normalize the vectors. By Normalize we mean creating the Unit Vectors along the same direction as the original two vectors
- Compute the magnitude of their dot product. That is, the dot product of the two normalized vectors

```python
# Import Numpy module
import numpy as np
```

```python
# Create two random-integer vectors
n = 4
v1 = np.round(20 * np.random.randn(n))
v2 = np.round(20 * np.random.randn(n))
```

```python
# Compute the Lengths of the individual vectors, and magnitude of their dot product
v1m = np.sqrt(np.dot(v1, v1))
# v1m = np.linalg.norm(v1)            # The norm() function calculate the length of the vector
v2m = np.sqrt(np.dot(v2, v2))
# v2m = np.linalg.norm(v2)
dpm = np.abs(np.dot(v1, v2))          # The abs() function is used to take the absolute value of dot product of v1 & v2
print(f'v1m = {v1m:.2f}')
print(f'v2m = {v2m:.2f}')
print(f'dpm = {dpm:.2f}')
```

```python
# Normalize the vectors
v1u = v1/v1m
v2u = v2/v2m
print(f"v1u = {v1u}")
print(f"v2u = {v2u}")
```

```python
# Verify that the magnitudes of the Normalized Vectors (Unit Vectors) are equal to 1
v1u_magnitude = np.linalg.norm(v1u)
v2u_magnitude = np.linalg.norm(v2u)
print(f"v1u_magnitude = {v1u_magnitude}")
print(f"v2u_magnitude = {v2u_magnitude}")
```

```python
# Compute the magnitude of the dot product of the two normalized vectors
normalized_dpm = np.abs(np.dot(v1u, v2u))
print(f"normalized_dpm = {normalized_dpm :.2f}")
```

Here, we observe that the magnitude of the dot product of the original, unscaled vectors is much larger than the magnitude of either individual vector. In contrast, for unit vectors, the magnitude of the dot product is smaller than or equal to the magnitude of either vector.<br>

Therefore, the dot product between two normalized vectors is an important quantity that provides information about the relationship between the two vectors. This concept is widely used in Statistics and Machine Learning, where it can help measure the similarity or alignment between vectors.
