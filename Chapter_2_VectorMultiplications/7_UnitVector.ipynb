### Unit Vector
A **unit vector** is a vector whose **magnitude (length)** is exactly 1, while its direction remains the same as the original vector. A unit vector is obtained by dividing a nonzero vector by its magnitude. This changes the vector’s length to 1 while preserving its original direction. For a vector **a**, the unit vector is:

$$
\hat{\mathbf{a}} = \frac{\mathbf{a}}{\|\mathbf{a}\|}
$$


```python
# Import Numpy & Matplotlib Modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a 2D Vector
v = np.array([-3, 6])

# Magnitude of Vector v
vMag = np.linalg.norm(v)

# Unit Vector
v_hat = v / vMag

# Plot both the Vector v and its Unit Vector v_hat
fig, axis = plt.subplots(figsize=(7, 7))

# Plot vector v
axis.quiver(
    0, 0, v[0], v[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='black',
    width=0.008,
    label=r'$\mathbf{v}$'
)

# Plot unit vector
axis.quiver(
    0, 0, v_hat[0], v_hat[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='green',
    width=0.008,
    label=r'$\hat{\mathbf{v}}$'
)

# Symmetric coordinate limits
axis.set_xlim(-7, 7)
axis.set_ylim(-7, 7)

# Equal x-y scaling
axis.set_aspect('equal')

# Remove the default rectangular box
axis.spines['top'].set_visible(False)
axis.spines['right'].set_visible(False)
axis.spines['left'].set_visible(False)
axis.spines['bottom'].set_visible(False)

# Draw x-axis through origin
axis.axhline(0, color='black', linewidth=1.2, zorder=0)
# Draw y-axis through origin
axis.axvline(0, color='black', linewidth=1.2, zorder=0)

# Add arrowheads to coordinate axes
axis.annotate(
    '',
    xy=(7, 0),
    xytext=(6.5, 0),
    arrowprops=dict(
        arrowstyle='->',
        color='black',
        linewidth=1.2
    )
)

axis.annotate(
    '',
    xy=(0, 7),
    xytext=(0, 6.5),
    arrowprops=dict(
        arrowstyle='->',
        color='black',
        linewidth=1.2
    )
)

# Add Grid
axis.grid(True, linestyle='--', alpha=0.2)

# Axis labels
axis.text( 6.7, -0.45, 'x', fontsize=14)
axis.text( 0.25, 6.7, 'y', fontsize=14)

# Origin
axis.scatter(0, 0, color='black', s=25, zorder=5)
axis.text(0.2, 0.2, '(0, 0)', fontsize=11)

# Vector labels
axis.text(
    v[0] - 0.45,
    v[1] + 0.15,
    r'$\mathbf{v}$',
    color='black',
    fontsize=16
)

axis.text(
    v_hat[0] + 0.15,
    v_hat[1] + 0.15,
    r'$\hat{\mathbf{v}}$',
    color='green',
    fontsize=16
)

# Add Legend
axis.legend( fontsize=12, loc='upper right')

# Title
axis.set_title(
    r'Vector $\mathbf{v}$ and Its Unit Vector $\hat{\mathbf{v}}$',
    fontsize=15
)

plt.show()
```


The above is the case when the magnitude of the original vector **v** is greater than 1.
