### Code Challenge - Dot Product with Matrix Columns
- Create 2 4x6 matrices of random numbers
- Use a for Loop to compute Dot Products between corresponding columns 

```python
import numpy as np
```

```python
A = np.random.randn(4,6)
B = np.random.randn(4,6)

print("A = ")
print(A)
print(' ')
print("B = ")
print(B)
```

```python
# Compute Dot Products between corresponding columns of A and B
dot_products = np.zeros(A.shape[1])
for column in range(A.shape[1]):
    # print("D"+str(column), "= ", np.dot(A[:,column], B[:, column]))
    dot_products[column] = np.dot(A[:,column], B[:, column])

print("dot_products = ")
print(dot_products)
```
