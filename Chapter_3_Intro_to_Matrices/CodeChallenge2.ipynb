### Code Challenge - Linearity of Trace
The goal of this code challenge is to explore the **trace** and in particular whether the **trace** is a linear operator. The operator is linear if it is closed under addition and scalar multiplication. In paractice that means that we want to explore two things:
- Determine the relationship between **tr(A) + tr(B)** and **tr(A + B)**. If the **trace** is a linear operator then these two quantities should be equal.
- Determine the relationship between **tr(s * A)** and **s * tr(A)**. Again if the **trace** is a linear operator then these two quantities should be equal.

Note that **Diagonal** is defined for both the **Square Matrices** as well as **Rectangular Matrices**, but the **trace** is defined only for **Square Matrices**.

```python
# Import Numpy module
import numpy as np
```

```python
# Define sizes of matrices
m = 4
n = m
```

```python
# Create two Square Matrices
A = np.random.randn(m, n)
B = np.random.randn(m, n)
```

```python
# Compute tr(A + B) and tr(A) + tr(B)
print(f"tr(A + B) = {np.trace(A + B)}")
print(f"tr(A) + tr(B) = {np.trace(A) + np.trace(B)}")
```

```python
# Create a scalar
s = np.random.randn()
```

```python
# Compute tr(s * A) and s * tr(A)
print(f"tr(s * A) = {np.trace(s * A)}")
print(f"s * tr(A) = {s * np.trace(A)}")
```

Thus we can say that the the **trace** is a linear opeartor
