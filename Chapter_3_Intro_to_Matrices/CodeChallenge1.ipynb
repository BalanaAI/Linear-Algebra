### Code Challenge - Is Matrix-Scalar Multiplication a Linear Operation
Test for some random MxN matrices whether **s(A + B) = sA + sB** 

```python
# Import Numpy module
import numpy as np
```

```python
# Create Matrix Sizes
m = 7
n = 5
```

```python
# Create two random matrices of m x n
A = np.random.randn(m, n)
B = np.random.randn(m, n)
```

```python
# Create a random scalar
s = np.random.randn()
```

```python
# Compute L.H.S of the Equation
resL = s * (A + B)
print(resL)
```

```python
# Compute R.H.S of the Equation
resR = s * A + s * B
print(resR)
```

```python
# Verify that L.H.S - R.H.S = 0 Matrix
print(resL - resR)
```

This conforms our hypothesis that matrix-scalar multiplication is linear. That is, it obeys that distributive law.
